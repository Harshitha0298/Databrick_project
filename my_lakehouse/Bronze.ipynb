#read from the csv source systems

```python
df = (
    spark.read.option("header", "true")
    .option("inferSchema", "true")
    .csv("/Volumes/workspace/bronze/source_systems/Source_crm/sales_details.csv")
)
df.display()
```

#Write to the bronze layer


```python
df.write.mode("overwrite").option("overwriteSchema", "true").saveAsTable("workspace.bronze.crm_sales_details")
```

```python
%sql
select * from workspace.bronze.crm_sales_details
```
