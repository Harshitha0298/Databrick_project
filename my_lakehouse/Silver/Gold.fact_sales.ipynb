```python
query = """
SELECT
    sd.order_number,
    pr.product_key,
    cu.customer_id,
    sd.order_date,
    sd.ship_date,
    sd.due_date,
    sd.sales,
    sd.quantity,
    sd.price
FROM silver.crm_sales sd
LEFT JOIN gold.dim_products pr
    ON sd.product_key = pr.product_key
LEFT JOIN gold.dim_customers cu
    ON sd.customer_id = cu.customer_id;
"""
df = spark.sql(query)

     
```

```python
df.display()
```

```python
df.write.mode("overwrite").saveAsTable("gold.fact_sales")
```
