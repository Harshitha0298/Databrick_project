```python
query="""
SELECT
    ROW_NUMBER() OVER (ORDER BY ci.customer_id) AS customer_num,
    ci.customer_id,
    ci.customer_key,
    ci.customer_firstname,
    ci.customer_lastname,
    la.country,
    ci.customer_marital_status,
    CASE
        WHEN ci.customer_gender <> 'n/a' THEN ci.customer_gender
        ELSE COALESCE(ci.customer_gender, 'n/a')
    END AS gender,
    ca.birth_date AS birthdate,
    ci.customer_create_date AS create_date
FROM silver.crm_customer ci
LEFT JOIN silver.cust_az12 ca
    ON ci.customer_key = ca.customer_id
LEFT JOIN silver.erp_loc la
    ON ci.customer_key = la.customer_id
    where ci.customer_id is not null
    """

df = spark.sql(query)
```

```python
df.display()
```

#writing to gold

```python
df.write.mode("overwrite").format("delta").saveAsTable("gold.dim_customers")
```
